## 🛡️ Casting a Basic Ward (Try-Except)

To protect ourselves, we wrap dangerous magic in a `try` block. If a wild surge happens, the `except` block catches it!


```
try:
    print("🔮 Attempting to divide mana by zero...")
    mana = 100
    power = mana / 0
    print("This line will never execute!")
except ZeroDivisionError:
    print("🛡️ Ward Activated! You cannot divide by zero in this realm!")
```

## 🔍 Identifying the Wild Surge (Catching specific errors)

Sometimes, we need to know exactly *what* went wrong. Is it a missing item? A wrong spell type? We can catch specific errors or grab the error message itself.


```
try:
    spell_scroll = int("Fireball") # Trying to turn a word into a number!
except ValueError as error_message:
    print(f"🧙‍♂️ Oops! A ValueError occurred. The Oracle says: {error_message}")
```

## 🧹 Cleaning Up (Finally)

Whether the spell succeeds or fails, sometimes you need to sweep up the magical residue. The `finally` block *always* runs, no matter what!


```
try:
    print("✨ Casting a risky spell...")
    # 10 / 0
    print("✨ Spell succeeded!")
except Exception:
    print("💥 The spell failed!")
finally:
    print("🧹 Cleaning up the ritual circle (This always prints!).")
```
