# 🧰 Topic 6: The Academy's Armory (Internal Packages)

Every wizard knows that you don't need to invent every spell from scratch. The Great Academy provides a vast armory of pre-written spellbooks called **Modules** or **Packages**. These are built right into Python!

Today, we will borrow books from the Academy's library, including the Tomes of `random`, `sys`, and `os`.

---

## 🎲 The Tome of Chance (Random)

Magic is unpredictable! The `random` module allows us to harness chaos and generate random numbers, pick random elements, or shuffle our decks.

To use a spellbook, we must `import` it.


```
import random

# Rolling a magical 20-sided die
d20_roll = random.randint(1, 20)
print(f"🎲 You rolled a {d20_roll}!")

# Picking a random potion from our satchel
potions = ["Healing", "Invisibility", "Strength", "Mana"]
chosen_potion = random.choice(potions)
print(f"🧪 You blindly grab a Potion of {chosen_potion}!")
```

## 🕰️ The Time Weaver (Time)

Sometimes a wizard must wait. The `time` module allows us to pause our spells.


```
import time

print("⏳ Brewing potion...")
# time.sleep(2) # Pauses the spell for 2 seconds (uncomment to test)
print("✨ Potion is ready!")
```

## 🗺️ The Cartographer (OS & Sys)

The `os` (Operating System) and `sys` (System) modules let us talk directly to the golem running our spells (our computer). We can explore files, check our surroundings, and even end the program completely.


```
import os
import sys

# Where in the realm are we?
current_location = os.getcwd()
print(f"🗺️ Our current working directory is: {current_location}")

# Which version of Python magic are we wielding?
print(f"🐍 Python Version: {sys.version.split()[0]}")
```
