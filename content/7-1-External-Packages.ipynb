# 🌍 Topic 7: Bazaars of the Far Realms (External Packages)

While the Academy's armory is vast, the world is even bigger! Millions of wizards across the globe share their custom spellbooks in the Great Bazaar known as **PyPI** (The Python Package Index).

To bring these into our realm, we usually use the summoning command `pip install <package_name>`. Once installed, we can import them just like internal packages.

---

## 🐄 Speaking with Beasts (Cowsay)

Let's start with a fun one. The `cowsay` package allows us to summon ASCII beasts to deliver our messages!
*(Note: In a real environment, you'd need to run `pip install cowsay` first, but we'll assume the Academy has provided it).*


```
# Assuming cowsay is installed in our environment
try:
    import cowsay
    cowsay.cow("Moo-gic is real, Initiates!")
    cowsay.dragon("BEHOLD MY FIERY BREATH!")
except ModuleNotFoundError:
    print("📜 The 'cowsay' scroll is missing! Summon it with 'pip install cowsay'")
```

## 🦅 Sending Owls (Requests)

The `requests` library is one of the most powerful spellbooks in existence. It allows your program to send messengers (requests) across the mystical Web (Internet) to fetch data from other realms (APIs).

Let's fetch a random joke from the internet!


```
try:
    import requests

    # Sending a messenger to the Joke Realm
    print("🦅 Sending an owl to fetch a joke...")
    response = requests.get("https://official-joke-api.appspot.com/random_joke")

    # If the messenger returns successfully (Status 200)
    if response.status_code == 200:
        joke_data = response.json() # Translating the message to a Python dictionary
        print(f"\nSetup: {joke_data['setup']}")
        print(f"Punchline: {joke_data['punchline']}")
    else:
        print("The owl got lost!")
except ModuleNotFoundError:
     print("📜 The 'requests' scroll is missing! Summon it with 'pip install requests'")
```
