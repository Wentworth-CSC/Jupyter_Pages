# 📜 Topic 9: Reading the Ancient Scrolls (File I/O)

Knowledge is power, but it is fleeting if not recorded. Wizards must know how to scribe their findings into **Scrolls** (Files) and read from ancient texts.

Today, we learn File Input/Output (I/O) to read and write text files.

---

## ✍️ Scribing a New Scroll (Writing to a File)

To write to a file, we use the `open()` spell with the `"w"` (write) mode.
**Warning:** `"w"` will overwrite existing scrolls! If you want to add to a scroll, use `"a"` (append).

We use the `with` keyword, which acts as a magical ward that automatically closes the scroll when we are done, even if an error occurs.


```
scroll_name = "secret_spell.txt"

# Open the file in Write ('w') mode
with open(scroll_name, "w") as scroll:
    scroll.write("Fireball Spell Instructions:\n")
    scroll.write("1. Gather sulfur and bat guano.\n")
    scroll.write("2. Speak the words: 'Ignis Maximus!'\n")

print(f"✍️ The scroll '{scroll_name}' has been inscribed!")
```

## 📖 Reading the Ancient Texts (Reading a File)

Now, let's read the spell back from the scroll. We use the `"r"` (read) mode.


```
print("📖 Opening the scroll to read...\n")

with open(scroll_name, "r") as scroll:
    # Read all lines into a list
    lines = scroll.readlines()

    # Loop through the lines and chant them
    for line in lines:
        print(f"🧙‍♂️ Chanting: {line.strip()}") # .strip() removes extra newlines
```

## ➕ Adding to the Archives (Appending)

Let's add a warning to the end of our spell scroll without erasing what's already there. We use the `"a"` (append) mode.


```
with open(scroll_name, "a") as scroll:
    scroll.write("WARNING: Stand back 30 feet before casting!\n")

print("➕ Warning appended to the scroll.")

# Let's read it one last time to see the whole file
with open(scroll_name, "r") as scroll:
    print("\n📜 Final Scroll Contents:")
    print(scroll.read()) # .read() reads the entire file as one giant string
```
