# ⚖️ Topic 8: The Proving Grounds (Unit Testing)

A wizard cannot simply cast a fireball and *hope* it works. Before taking a spell into battle, it must be tested in the Proving Grounds!

**Unit Testing** is the practice of writing small, automated spells that test your main spells to ensure they behave exactly as expected. We will explore both the built-in `unittest` tome and the popular `pytest` framework.

---

## 🧪 Our Spell to Test

First, let's create a simple spell (function) that we want to test. It calculates the damage of a magical attack.


```
def calculate_damage(base_power, multiplier):
    """Calculates total magical damage."""
    if multiplier < 0:
        raise ValueError("Multiplier cannot be negative!")
    return base_power * multiplier

# Let's try it manually
print(f"Base 10, Multiplier 2: Damage = {calculate_damage(10, 2)}")
```

## 🏛️ The Built-in Proving Ground (Unittest)

Python comes with `unittest`. It requires us to create a "Test Case" class. It's formal and structured, like a classic duel.


```
import unittest

class TestMagicDamage(unittest.TestCase):

    def test_normal_damage(self):
        # We assert that 10 * 3 should equal 30
        self.assertEqual(calculate_damage(10, 3), 30)

    def test_zero_multiplier(self):
        self.assertEqual(calculate_damage(50, 0), 0)

    def test_negative_multiplier(self):
        # We assert that a ValueError is raised if the multiplier is negative
        with self.assertRaises(ValueError):
            calculate_damage(10, -1)

# In a normal script, we would run unittest.main().
# In Jupyter, we can run the tests like this:
print("⚖️ Running Unittest Proving Grounds...")
suite = unittest.TestLoader().loadTestsFromTestCase(TestMagicDamage)
unittest.TextTestRunner(verbosity=2).run(suite)
```

## 🚀 The Modern Arena (Pytest)

Many wizards prefer `pytest`. It is an external package (requires `pip install pytest`) and is much more readable. You don't need classes, just simple `assert` statements!

*(Note: Pytest is usually run from the command line, but we can see how the code looks here).*


```
# Pytest code example (Do not run here directly without pytest installed)

def test_pytest_normal_damage():
    assert calculate_damage(10, 3) == 30

def test_pytest_zero():
    assert calculate_damage(50, 0) == 0

print("✅ Pytest spells written! Run them in the terminal using the 'pytest' command.")
```
